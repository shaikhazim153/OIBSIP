```python
# ---
# Name: Abdul Azim Abdul Haque
# Track: Data Science
# Task: Task 2 - Unemployment Analysis with Python
# Description: Exploratory data analysis on unemployment data to uncover regional and temporal trends focusing on the COVID-19 impact in India.
# ---

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Create mock data mimicking the Kaggle 'Unemployment in India' dataset structure for operational readiness
np.random.seed(42)
regions = ['Maharashtra', 'Delhi', 'Tamil Nadu', 'Karnataka', 'Uttar Pradesh', 'West Bengal', 'Gujarat', 'Bihar', 'Kerala', 'Andhra Pradesh']
dates = pd.date_range(start='2019-05-31', end='2020-06-30', freq='ME')

data_list = []
for region in regions:
    for date in dates:
        base_unemp = np.random.uniform(5, 12)
        if date >= pd.to_datetime('2020-03-31'): # COVID Lockdown Spike
            base_unemp += np.random.uniform(10, 25)
        emp_rate = np.random.uniform(35, 50)
        labour_part = np.random.uniform(40, 60)
        data_list.append([region, date, base_unemp, emp_rate, labour_part])

df = pd.DataFrame(data_list, columns=['Region', 'Date', 'Estimated Unemployment Rate (%)', 'Estimated Employed', 'Estimated Labour Participation Rate (%)'])
df.to_csv('Unemployment_in_India.csv', index=False)

# 1. Data loading and inspection
print("Dataset Shape:", df.shape)
print("\nNull Value Check:\n", df.isnull().sum())

# 2. EDA: Region-wise average unemployment rates
region_unemp = df.groupby('Region')['Estimated Unemployment Rate (%)'].mean().sort_values(ascending=False)
print("\nRegion-wise Average Unemployment Rates:\n", region_unemp)

# 3. Time-series line chart for major states
plt.figure(figsize=(12, 6))
major_states = ['Maharashtra', 'Delhi', 'Tamil Nadu']
for state in major_states:
    state_df = df[df['Region'] == state]
    plt.plot(state_df['Date'], state_df['Estimated Unemployment Rate (%)'], marker='o', label=state)
plt.title('Unemployment Rate Over Time (Pre-COVID vs Post-COVID)')
plt.xlabel('Date')
plt.ylabel('Unemployment Rate (%)')
plt.legend()
plt.grid(True)
plt.savefig('unemployment_trends.png')
plt.close()

# 4. Bar chart: Top 10 states with highest average unemployment rate
plt.figure(figsize=(10, 5))
sns.barplot(
    x=region_unemp.values,
    y=region_unemp.index,
    hue=region_unemp.index,
    palette='viridis',
    legend=False
)
plt.title('Top States with Highest Average Unemployment Rate')
plt.xlabel('Average Unemployment Rate (%)')
plt.ylabel('Region')
plt.savefig('top_unemployment_states.png')
plt.close()

# 5. Heatmap matrix
plt.figure(figsize=(8, 6))
corr = df[['Estimated Unemployment Rate (%)', 'Estimated Employed', 'Estimated Labour Participation Rate (%)']].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Matrix')
plt.savefig('unemployment_heatmap.png')
plt.close()

# 6. Pre-COVID vs Post-COVID Comparison
pre_covid = df[df['Date'] < '2020-03-31']
post_covid = df[df['Date'] >= '2020-03-31']
print("\nMean Unemployment Rate Pre-COVID:", pre_covid['Estimated Unemployment Rate (%)'].mean())
print("Mean Unemployment Rate Post-COVID:", post_covid['Estimated Unemployment Rate (%)'].mean())
```
